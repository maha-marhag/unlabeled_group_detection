# Stage Identification Analysis

This notebook demonstrates the implementation in `stage_identification_analysis.py`. It keeps the implementation as the source of truth and only runs the script, then previews the main output tables.

```python
from pathlib import Path
import pandas as pd
import subprocess

project_root = Path.cwd().parents[1] if Path.cwd().name == 'graph-matching' else Path.cwd()
script = project_root / 'code' / 'graph-matching' / 'stage_identification_analysis.py'
output_dir = project_root / 'code' / 'graph-matching' / 'outputs' / 'stage_analysis'

subprocess.run(['python3', str(script), '--approach', 'overlap'], check=True)
```

## `(u,g,t)` Membership Structure

The full table has one row per user, final labeled group, and snapshot.

```python
sizes = pd.read_csv(output_dir / 'overlap' / 'labeled_group_sizes_by_timestamp.csv')
sizes.pivot(index='identity_group', columns='snapshot_index', values='member_count')
```

## Classic vs. Event-Based Stages

```python
classic = pd.read_csv(output_dir / 'overlap' / 'classic_stage_segments.csv')
event_based = pd.read_csv(output_dir / 'overlap' / 'event_based_stage_segments.csv')

pd.concat([
    classic.assign(source='classic'),
    event_based.assign(source='event_based'),
]).groupby(['source', 'stage_type']).size().unstack(fill_value=0)
```
